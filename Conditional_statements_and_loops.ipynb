# Conditional Statements & Loops
These are control flow mechanisms in python

## Conditional Statements
Conditional statements test whether a specific condition is TRUE or FALSE using comparison(==, !=, <, >, <=, >=) and logical(and, or, not) operators.

Key syntax
- if: executes a block if condition is TRUE
- elif (short for else if): Checks another condition if prior conditions are FALSE.
- else: Executes if none of the prior conditions are met

```python
# If statement
marks = 72
if marks >= 50:
    print("pass")
```

```python
marks = 45
if marks >= 50:
    print("pass")
else: 
    print("Fail")
```

```python
# Example: Grading
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >=70:
    print("Grade: C")
else:
    print("Grade: D") 
```

### Nested Conditional statements
A nested if statement is an if, elif, or else block placed inside another if, elif, or else block.
- Python uses 4-space indentation levels to define which block a nested condition belongs to.
- Outer conditions are evaluated first, inner conditions are checked only if the outer conditions are TRUE.

```python
income = 55000
credit_score = 720
has_existing_loan = False

if income >= 50000:
    print("Income requirement met.")
    # Outer condition passed, now checking nested conditions
    if credit_score >= 700:
        if not has_existing_loan:
            print("Status: Approved for Tier-1 Preferred Rate")
        else:
            print("Status: Approved for Standard Rate (Existing Loan Present)")
    else:
        print("Status: Conditional Approval (Credit Score Below 700)")
else:
    print("Status: Rejected (Income Below 50,000)")
```

```python
score = 78
attendance = 80

if score >= 90:
    if attendance >=90:
        print("Permitted to sit exam")
else: 
    print("Not cleared")
```

## Loops
Loops repeat a block of code multiple times.
Two kinds of loops
- For loops
- while loop

```python
# Iterating using range
for i in range(5):
    print(i)
```

### For loop
- Used when iterating over a sequence(such as a list, string, or range)
- used when number of iterations is known

```python
for _ in range(1,6): #(start,stop)
    print(_)
```

```python
for i in range(2,6,2):
    print(i)
```

```python
#Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

### While loop
- Repeats code as long as specified condition remains TRUE.
- Requires a manual control(like incrementing a counter) to prevent infinite loops.


```python
count = 3
while count > 0:
    print(count)
    count -= 1
```

```python
# Example
count = 1

while count < 5:
    print(count)
    count += 1  # incrementing counter
```

### Nested Loops
A nested loop is a loop inside another loop.
- Outer loop controls the higher level iteration
- Inner loop completes its entire cycle for every single iterationof the outer loop.
    

```python
x = 1
while x <= 2:
    y = 1
    # Inner loop resets and completes for every step of x
    while y <= 3:
        print(f"Point: ({x}, {y})")
        y += 1
    x += 1
```

```python
for i in range(1,4):
    for j in range(1,4):
        print(i * j, end=" ") """ for the first iteration of the outer loop, 
                                1 is multiplied by values 1,2,3."""
    print()
```

### Combining nested conditionals and loops
Enables filtering, searching data

```python
fruits = ["apple", "orange", "Kiwi", "Mango"]
my_fruit = "Orange"
for fruit in fruits:
    if fruit == my_fruit:
        print("Got it")
    else:
        print("Not here")
    
```

### Loop control statements
Modify the normal execution flow inside loops
- break: exits the loop immediately
- continue: skips the current iteration and jumps to the next one

```python
# Demonstrating break and continue
numbers = [1, 2, 3, 4, 5, 6]

for num in numbers:
    if num % 2 == 0:
        continue  # Skip even numbers
    if num == 5:
        break     # Stop loop when 5 is reached
    print(num)
```

```python
for i in range(5):
    print(i)
    if i == 2:
        break
```

```python
for i in range(5):
    if i == 2 or i == 4:
        continue # skips 2 and 4
    print(i)
```

```python
scores = [45, 68, 82, 90, 39, 55]
passed = 0
for score in scores:
    if score >= 50:
        passed += 1
print("Passed:", passed)
```

```python
#for a conditional inside a loop
scores = [45, 68, 82, 90, 39, 55]
for score in scores:
    if score >= 50:
        print("passed")
    else: 
        print("below pass mark")
```

```python
# for a loop inside a conditional
scores = [45, 68, 82]
if len(scores) > 0:
    total = 0
    for s in scores:
        total += s
    print("Average:", total / len(scores))
else:
    print("No scores to process")
```
