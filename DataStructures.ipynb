# Lists, Tuples, Sets, and Dictionaries (Collection data types)

## Lists

- A list is an array of ordered values
- List is a built-in data structure used to store an ordered collection of items.
- Ordered here means keeping your position allocated implying fixed positions
### Characteristics of lists
- Values are enclosed in square brackets[] and separated by commas
- Can store more than one data type
- Mutable ie elements can be changed, updated, added or removed
- Elements main the order in which they are inserted
- holds duplicates

```python
num = [2,4,7,1]

print(num)

```

```python
type(num)
```

```python
len(num)
```

```python
2 in num
```

```python
num[1]
```

```python
num.append(8)
print(num)
```

```python
num.append([7,6])
print(num)
```

```python
num.extend([3,5])
print(num)
```

```python
num.insert(3,5)
print(num)
```

```python
num.count(5)
```

```python
fruits = ["Apple", "Orange", "Banana", "Grapes", "Apple"]
print(fruits)
```

## Sets

Set is a built in python data type used to store a collection of unique items
### xtics
- Stores only unique elements: duplicates are removed
- Unordered; elements do not have a fixed position and cannot be accessed using indexes
- elements are enclosed in curly brackets{} and separated by commas
- sets do not maintain insertion order

```python
fruits = {"Mango", "Orange", "Kiwi", "Apple","Kiwi"}
print(fruits)
     
```

**There are two types of sets**
- Sets: They are mutable and new elements can be added once sets are defined
- Frozen Sets: They are immutable and new elements cannot be added fter its defined

```python
cities = {"Kampala","Jinja","Gulu","Mbarara","Mbale"}
print(cities)
cities.add("Fort Portal")
print(cities)
```

## Dictionary

- Dictionary is a collection of key-value pairs.
- Dictionary is surrounded by curly brackets{}
- Each pair is separated by a comma, and the key and value are separated by a colon:
- Dictionary is mutable
- Each key is unique, it points to a value that can be any object

```python
state_capital = {"Uganda": "Kampala",
                 "Kenya": "Nairobi",
                 "Tanzania": "Dodoma",
                 "Rwanda": "Kigali",
                 "Burundi": "Bujumbura"
                }
print(state_capital)
```

```python
print(state_capital["Rwanda"])
```

```python
state_capital.get("Rwanda")
```

```python
state_capital["DRC"] = "Kinshasa"
print(state_capital)
```

```python
"DRC" in state_capital
```
