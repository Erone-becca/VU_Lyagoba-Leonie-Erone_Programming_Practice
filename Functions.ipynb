```python
# Recursion
def factorial(n):
    if n==1:
        return 1
    return n * factorial(n-1)

factorial(4)
```

```python
# tail and non-tail recursion
# tail recursion
def tail_fact(n, acc=1):
    if n == 0:
        return acc
    else:
        return tail_fact(n-1, acc * n)

tail_fact(5)
```

```python
# non- tail recursion
def non_tail_fact(n):
    if n == 0:
        return 1
    else:
        return n * non_tail_fact(n-1)

print(non_tail_fact(5))
    
```

```python
# recursion
def write_animals(n=0):
    animals = ["lion", "tiger", "zebra"]
    if n < len(animals):
        print(animals[n], end=" ") # end="" makes output appear horizontally
        return write_animals(n + 1)
    else:
        return

write_animals()
```
