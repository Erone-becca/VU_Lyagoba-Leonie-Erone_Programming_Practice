```python
#Histogram showing wine quality distribution
plt.hist(df['quality'], bins=7, color='skyblue', edgecolor='black')
plt.title('Wine Quality Distribution')
plt.xlabel('Quality Score')
plt.ylabel('Count')
plt.show()
```

```python
#Scatter plot
plt.scatter(df['alcohol'], df['density'], alpha=0.4, color='coral')
plt.title('Alcohol vs. Density')
plt.xlabel('Alcohol (%)')
plt.ylabel('Density')
```

```python
#barchart
wine_counts = df['wine_type'].value_counts()
plt.bar(wine_counts.index, wine_counts.values, color=['lightgreen', 'salmon'], edgecolor='black')
plt.title('Count by Wine Type')
plt.ylabel('Count')
```

```python
#Scatterplot of fixed acidity vs pH
plt.scatter(df['fixed_acidity'], df['ph'], alpha=0.4, color='mediumpurple')
plt.title('Fixed Acidity vs. pH')
plt.xlabel('Fixed Acidity')
plt.ylabel('pH')
```
